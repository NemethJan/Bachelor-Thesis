**Constructing the Y-Variable**

This brings the thesis to the first data handling task. In the previous step, the Compustat database was loaded. However, raw features like EBITDA or Sales do not represent the specific growth definition needed for this thesis. To construct the real growth rates, the first step is to see how many real observations are available in the specified Target Variable (Sale, EBITDA, FCF) and to construct FCFs first. This has to be done now, because the results should be independent of any downstream filtering decision later.

To calculate Free Cashflow to the Firm (FCFF) the proven formula is used:

**Operating Cash Flow (oancf) + Interest Expense (xint) * (1 - Tax Rate) - Capital Expenditures (capx)**

Every parameter is available in the dataset except for the tax rate. However, the tax rate can be calculated from the dataset using Tax Expenditures (txt) divided by Pretax Income (pi). The formula is as follows: Tax Rate = txt / pi. Furthermore, the tax rate has to be clipped at 0. A negative tax rate is not possible, while negative tax expenditures / negative pretax income are possible. Furthermore, the peak U.S. tax rate since 1987 was 40%, further values do not make any sense, so this will be the upper boundary: https://taxpolicycenter.org/sites/default/files/statistics/pdf/corporate_top_rate_bracket_1.pdf

```python
  
# Y-VARIABLE LOOKUPS


#Define a subset with relevant columns needed for the Y-Variable
funda_y = compustat_funda.reset_index()[
    ['gvkey', 'fyear', 'oibdp', 'sale', 'oancf', 'capx']].copy()

funda_y['fyear'] = pd.to_numeric(funda_y['fyear'], errors='coerce')


# FCF = Operating Cash Flow − Capex (no tax/interest adjustment)
funda_y['fcff'] = funda_y['oancf'] - funda_y['capx']

#Build a function to efficiently look at how many observations I have in each target
def build_lookup(df, col):
    return (df[['gvkey', 'fyear', col]]
            .dropna(subset=['fyear', col])
            .drop_duplicates(['gvkey', 'fyear'], keep='last')
            .set_index(['gvkey', 'fyear'])[col])


y_lookups = {'ebitda': build_lookup(funda_y, 'oibdp'),
               'sale':   build_lookup(funda_y, 'sale'),
               'fcff':   build_lookup(funda_y, 'fcff')}

#Extract the look up sizes (Keys: Target Variable, Values: the Nmber of Observations)
{k: len(v) for k, v in y_lookups.items()}  
```

Sale has the most non-missing observations with 269884, followed by EBITDA with 262527 and FCFF with 248418.

In a next step, the first feature discussed in the thesis (lagged time-series growth) is built. Because of the potentially high missingness in a time-series feature (5 years are needed for one growth calculation), a fallback decision is implemented, using year 4 when year 5 is not available and year 3 when year 5 and 4 are not available and so on.

Again, this has to be done here so that no further downstream cleaning decisions affect the feature building and the full observation set possible can be used.

```python
#Initialize an Empty Lagged Growth Dataframe 
lag_growth = pd.DataFrame()


#Iterate trough the target variabls 
for var in ['fcff', 'oibdp', 'sale']:
    # Create a wide table with one obserrvation per firm and one column per year, with the Values representing the itaterated target elements 
    wide = funda_y.pivot_table(index='gvkey', columns='fyear', values=var, aggfunc='last')

    #Initialize an empty growth dataframe with the same wide structure as the wide datafram 
    growth = pd.DataFrame(np.nan, index=wide.index, columns=wide.columns)

    # Iteratre trough the  years 5 to 1
    for k in [5, 4, 3, 2, 1]:
        #Shift the value using the value from k prior years
        lag = wide.shift(k, axis=1)              
        #Calculate the growth rate
        g_k = (wide / lag) ** (1/k) - 1      

        #Lag 0 not computable 
        g_k = g_k.where(lag > 0)

        #Fill the growth set with the g_k estimates
        #Notice that the column will only be filled if it has missing Values. If its filled by a year 5 lag it will not be overwritten by a year 1 lag            
        growth = growth.fillna(g_k)              


    # Not I add the calcualted growth back into the lagged growth dataset and convert it to long format 
    lag_growth[var + '_growth_lag'] = growth.stack() 

lag_growth = lag_growth.reset_index()  
```

### 2.2. CRSP: Return x Delisting Data

**Creating the crsp_merged dataset**

In a next step, the CRSP Return Dataset is merged with the CRSP Delisting Dataset to mitigate the survivorship bias mentioned in the thesis. The unique identifiers for the merge are the PERMNO and the date. Since both datasets are from CRSP, there should not be any big issues with the matching. The return dataset is used as the left dataset in a left join.

Next, for all the applicable observations, the return is adjusted by the delisting return mentioned in detail below. After that, the impact of the delisting return adjustment on the return distribution is examined.

```python
#Convert both tables to datetime format for the date columns:
crsp_returns['MthCalDt'] = pd.to_datetime(crsp_returns['MthCalDt'])
crsp_delisting['DelistingDt'] = pd.to_datetime(crsp_delisting['DelistingDt'])

#Rename the PERMNO column in crsp_returns to match the crsp_delisting dataset for merging purposes
crsp_returns.rename(columns={'PERMNO': 'permno'}, inplace=True)

#Form Year month columns for both tables:
crsp_returns['year_month'] = crsp_returns['MthCalDt'].dt.to_period('M')
crsp_delisting['year_month'] = crsp_delisting['DelistingDt'].dt.to_period('M')
```

```python
#Merge the two tables on permno and year_month to get the delisting information into the returns table:
crsp_merged = crsp_returns.merge(
    crsp_delisting[['permno', 'year_month', 'DelRet', 'DelReasonType', 'DelRetMissType']],
    on=['permno', 'year_month'],
    how='left'
)
```

**Adjusting the return column by the delisting return**

Unfortunately, the delisting reasons are coded in a way that is not very intuitive. Below is a dictionary that translates the codes into more understandable descriptions. Not every delisting has a delisting return attached to it; to follow existing literature (Shumway 1997), the delisting return is set to -30% if the delisting occurs because of one of the following factors and the delisting return is not available (this affects roughly 1.424 out of 23230 delisting rows).

```python

#Create a dictionary to translate the relevant delisting reason codes into more understandable descriptions
performance_reasons_dict = {
    'INSC': 'Insufficient capital / financial distress',
    'INSF': 'Insufficient public float',
    'FING': 'Financial/regulatory guidelines not met',
    'BKPY': 'Bankruptcy filing',
    'DELQ': 'Delinquent in filing / non-compliance',
    'EQRQ': 'Equity requirements not met',
    'SERQ': 'SEC regulatory requirements not met',
    'VIO':  'Violation of exchange rules',
    'DERE': 'Deregistration / SEC reporting failure'
}
#Create a list of the performance reasons for easier filtering
performance_reasons = list(performance_reasons_dict.keys())

#Needs Imputation: Identify rows where DelRet is missing and the DelReasonType is in one of the performance reasons in the before creatted list
needs_imputation = (crsp_merged['DelRet'].isna()) & (crsp_merged['DelReasonType'].isin(performance_reasons))

#Apply the classical boolean indexing to set DelRet to -0.3 for the identified rows
crsp_merged.loc[needs_imputation, 'DelRet'] = -0.3
```

```python
#Adjust Market returns for delisting returns (if applicable for a row): 

#Initialize the adjusted return column with the original returns
crsp_merged['mthret_adj'] = crsp_merged['MthRet']


#1. Case: Has delret and mthret notna
mask1 = crsp_merged['DelRet'].notna() & crsp_merged['MthRet'].notna()
crsp_merged.loc[mask1, 'mthret_adj'] = (
    (1 + crsp_merged.loc[mask1, 'MthRet']) *
    (1 + crsp_merged.loc[mask1, 'DelRet']) - 1
)

#2. Case: Has delret but mthret is na
mask2 = crsp_merged['DelRet'].notna() & crsp_merged['MthRet'].isna()
crsp_merged.loc[mask2, 'mthret_adj'] = crsp_merged.loc[mask2, 'DelRet']
```

### 2.3. Factor x CRSP

In this section, the Factor Dataset is merged with the previously merged CRSP Dataset. After the merge is done, the relevant factor columns are constructed from information in the merged dataset.

Specifically, the factors Beta, Idiosyncratic Volatility and Momentum are constructed here with the following formulas (can be found more clearly in the thesis Appendix):

***Beta: mthret_adj = α + β × mktrf + ε***

***Idiosyncratic Volatility = ε = mthret_adj - α - β × mktrf***

***IdioVol = annualized std(ε) = std(ε) × √12***

```python
#Convert to year_month format
wrds_factors['dateff'] = pd.to_datetime(wrds_factors['dateff'])
wrds_factors['year_month'] = wrds_factors['dateff'].dt.to_period('M')
```

```python
#Merge the factors with the CRSP dataset on year_month
crsp_merged = crsp_merged.merge(
    wrds_factors[['year_month', 'mktrf', 'rf']],
    on='year_month',
    how='left'
)
```

```python
#Construct the excess return column by subtracting the risk-free rate from the adjusted monthly return for the later beta and idiosyncratic volatility calculations
crsp_merged['excess_ret'] = crsp_merged['mthret_adj'] - crsp_merged['rf']
print(crsp_merged['excess_ret'].describe())
```

#### 2.3.1. Compute the Momentum Factor

The classical momentum factor from Jegadeesh and Titman (1993) is constructed by calculating the cumulative return over the last 12 months of a stock, excluding the most recent month (because of the potential short-term reversal effect). This thesis follows the classical definition. To calculate the momentum factor, one has to use the pandas groupby function to group the data by PERMNO and then apply a rolling window of 12 months to calculate the cumulative return (pandas method rolling). The most recent month is excluded from the calculation to avoid the short-term reversal effect by shifting the rolling window by 1 month.

Note that, as a data handling decision, the momentum factor is only calculated for stocks that have at least 9 months of return data available in the last 12 months. This is to ensure that the momentum factor is not calculated based on too few observations, which could lead to unreliable results. Other stocks that do not have enough return data are assigned a missing value for the momentum factor.

```python
# Calculate Momentum (t-12, t-1 Momentum) #Jegadeesh–Titman

#I use log returns because its simpler to calculate cumulative returns over multiple periods (summing)
crsp_merged['log_ret'] = np.log1p(crsp_merged['mthret_adj'].clip(lower=-0.9999)) 


#Construct the new momentum factor column by grouping the data by PERMNO, applying a rolling window of 11 months to calculate the cumulative return from t-12 until t-2, and then shifting the result by 2 months to exclude the most recent month and the last full month.
crsp_merged['mom_12_1'] = (
    crsp_merged.groupby('permno')['log_ret']
    .transform(lambda x: x.rolling(window=11, min_periods=9).sum()) #t-12 until t-2
    .groupby(crsp_merged['permno']).shift(2) #Skip the recent month (data leakage) and skip the last full month to get jegendesh momentum
)
#Finally, I exponentiate the result to convert it back from log returns to regular returns.
crsp_merged['mom_12_1'] = np.expm1(crsp_merged['mom_12_1'])
```

#### 2.3.2. Compute Beta & Idiosyncratic Volatility Factors

In the next section, the beta and idiosyncratic volatility factors are calculated. The beta factor is a measure of a stock's volatility in relation to the overall market. On the other hand, idiosyncratic volatility measures the volatility of a stock that is not explained by market movements. The following data handling decision is made: the beta factor is calculated over a 5 year rolling window, which is a common practice in finance. This means that for each month, the beta factor is calculated using the past 60 months of return data. If there are not enough data points available (less than 18 months), the beta factor is assigned a missing value. This ensures that the beta factor is based on a sufficient amount of historical data to provide a reliable estimate of the stock's volatility relative to the market.

Note: calculating the beta & idiosyncratic factors was significantly helped by Claude Opus 4.7.

```python
#Define the parameters for the rolling window regression of the Beta & Idiosyncratic Volatility factors
window = 60      
min_nobs = 18  
```

```python

DATE_COL = 'MthCalDt'

#Prepare the dataset: drop all rows with a missing excess return or market factor and sort by permno and date so the rolling windows run in the correct time order
df = crsp_merged.copy()
valid_mask = df[['excess_ret', 'mktrf']].notna().all(axis=1)
df = df[valid_mask].sort_values(['permno', DATE_COL]).reset_index(drop=True)

#Create the helper columns that are needed for the rolling sums of the regression
df['_x']  = df['mktrf']
df['_y']  = df['excess_ret']
df['_xx'] = df['mktrf'] * df['mktrf']
df['_yy'] = df['excess_ret'] * df['excess_ret']
df['_xy'] = df['mktrf'] * df['excess_ret']

#Group by permno so that every rolling window stays within one single stock
gb = df.groupby('permno', sort=False)

#Build a function to efficiently compute the rolling sum of a helper column (vectorised, because looping over every single stock would be far too slow on this sample size)
def rsum(col):
    return gb[col].transform(lambda s: s.rolling(window, min_periods=min_nobs).sum())

#Count the valid observations per window and compute the rolling sums for all helper columns
n   = gb['_y'].transform(lambda s: s.rolling(window, min_periods=min_nobs).count())
Sx  = rsum('_x')
Sy  = rsum('_y')
Sxx = rsum('_xx')
Syy = rsum('_yy')
Sxy = rsum('_xy')

#Center the moments around the window mean to get the sums of squares and the cross product needed for the Beta estimate
Sxx_c = Sxx - Sx * Sx / n      #Σ(x-x̄)²
Syy_c = Syy - Sy * Sy / n      #Σ(y-ȳ)²
Sxy_c = Sxy - Sx * Sy / n      #Σ(x-x̄)(y-ȳ)

#Set windows without any market variance to missing, since a Beta can not be estimated in this case
Sxx_c = Sxx_c.where(Sxx_c.abs() > 1e-10)

#Estimate the market model in closed form instead of running an actual regression: Beta = Sxy_c / Sxx_c and Alpha = ȳ - Beta * x̄
df['beta']  = Sxy_c / Sxx_c
df['alpha'] = Sy / n - df['beta'] * (Sx / n)

#Compute the residual of the single observation (the part of the excess return that is not explained by the market), kept identical to the previous version of this calculation
df['resid'] = df['excess_ret'] - df['alpha'] - df['beta'] * df['mktrf']

#Idiosyncratic Volatility is the annualised standard deviation of the residuals of the exact same 60 month window
#The sum of the squared residuals does not have to be computed observation by observation but follows directly from the centered moments: SSR = Syy_c - Beta * Sxy_c
ssr = (Syy_c - df['beta'] * Sxy_c).clip(lower=0)   #Clip at 0 to catch small negative values coming from floating point noise
df['idio_vol'] = np.sqrt(ssr / (n - 2)) * np.sqrt(12)   #Divide by n-2 because Alpha and Beta are estimated, then annualise the monthly standard deviation with √12

#Drop the helper columns again and store the final dataset including both new factors
df = df.drop(columns=['_x', '_y', '_xx', '_yy', '_xy'])
crsp_with_beta = df

#Sanity Check: look at the shape of the dataset and the missingness of both newly created factors
print("Shape:", crsp_with_beta.shape)
print(f"Missing Beta:    {crsp_with_beta['beta'].isna().mean()*100:.1f}%")
print(f"Missing IdioVol: {crsp_with_beta['idio_vol'].isna().mean()*100:.1f}%")
```

One can see that roughly 15 percent of the beta and idiosyncratic volatility values are missing, which is a significant proportion. This will be analysed below after looking at the distribution of the beta factor.

```python
#Create a year column for plotting purposes
crsp_with_beta['year'] = crsp_with_beta['MthCalDt'].dt.year

#Initialize a figure with two subplots for Beta and Idiosyncratic Volatility distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 10))
fig.suptitle('Beta & Idiosyncratic Volatility Sanity Checks', 
             fontsize=13, fontweight='bold')

#On the first axis plot the histogram of Beta values, clipping extreme values for better visualization
axes[0].hist(crsp_with_beta['beta'].dropna().clip(-3, 5), 
             bins=100, color='steelblue', edgecolor='white')
axes[0].set_title('Beta Distribution (clipped)')
axes[0].axvline(1, color='red', linestyle='--', label='Beta = 1')
axes[0].axvline(crsp_with_beta['beta'].median(), 
                color='orange', linestyle='--', 
                  label=f"Median = {crsp_with_beta['beta'].median():.2f}")
axes[0].legend()

#On the second axis plot the histogram of Idiosyncratic Volatility values, clipping extreme values for better visualization
axes[1].hist(crsp_with_beta['idio_vol'].dropna().clip(0, 2), 
               bins=100, color='firebrick', edgecolor='white')
axes[1].set_title('Idiosyncratic Volatility Distribution (clipped)')
axes[1].axvline(crsp_with_beta['idio_vol'].median(), 
                  color='orange', linestyle='--',
                  label=f"Median = {crsp_with_beta['idio_vol'].median():.2f}")
axes[1].legend()


plt.tight_layout()
plt.show()
```

One can see that the beta factor is roughly normally distributed with a mean of 1. This is consistent with the fact that the beta factor is a measure of a stock's volatility in relation to the overall market, and overall stocks are expected to replicate the market, resulting in a beta factor close to 1. The idiosyncratic volatility is right-skewed, which is consistent with the fact that most stocks have low idiosyncratic volatility, but a few stocks have very high idiosyncratic volatility.

### 2.5. Final Merge: Compustat x CRSP 

In this section the final merged datasets are created: Compustat is the main dataset and the most recent monthly market features (beta, vola, monthly return) that were available BEFORE the filedate for a given firm-year pair are needed, to avoid any time-leakage concerns.

The logic is that a machine model makes a forecast when the 10K is available and looks at the most recent available market data until then. Hence filedate >= Monthly Return Date. To implement this, a pandas as-of join is used again, but with Compustat as the left dataset now, again using direction backward. Different to the last section, the tolerance window is set to 40 days to ensure that the market features are within the last 40 days before the prediction date.

As mentioned before, two datasets are created: one with analyst coverage and one without analyst coverage. But first the Compustat dataset has to be merged with the Linkage Table between Compustat and CRSP to get the PERMNO for each gvkey.

```python
# Compustat × CRSP-Compustat Linkage Copy
link_cstat = compustat_crsp_link.copy()


#Convert the date columns to datetime format and handle the 'E' values in LINKENDDT by replacing them with a far future date as before 
link_cstat['LINKDT'] = pd.to_datetime(link_cstat['LINKDT'])
link_cstat['LINKENDDT'] = link_cstat['LINKENDDT'].replace('E', '2099-12-31')
link_cstat['LINKENDDT'] = pd.to_datetime(link_cstat['LINKENDDT'])

#Ensure that the datadate column in compustat_merged is also in datetime format
compustat_merged['datadate'] = pd.to_datetime(compustat_merged['datadate'])

#Ensure that the LPERMNO column is of integer type for merging purposes
link_cstat['LPERMNO'] = link_cstat['LPERMNO'].astype(int)


#I only keep the relevant columns for the merge
link_cstat = link_cstat[['gvkey', 'LPERMNO', 'LINKDT', 'LINKENDDT']].rename(
    columns={'LPERMNO': 'permno'}
)


# Compustat_merged x Linkage Merge
#Inner Join on gvkey to keep only the observations that have a corresponding link in the linkage table
compustat_merged = compustat_merged.merge(link_cstat, on='gvkey', how='inner')

print(f"\nCompustat before Merge: {len(compustat_merged):,}")

#Ensure robustness: The datadate columns has to be between the link start and end date for the merge to be valid.
compustat_merged = compustat_merged[
    (compustat_merged['datadate'] >= compustat_merged['LINKDT']) &
    (compustat_merged['datadate'] <= compustat_merged['LINKENDDT'])
].copy()


print(f"\nCompustat after the Merge & Robustness: {len(compustat_merged):,}")

# Drop columns that are no longer needed after the merge to clean up the dataset
compustat_merged = compustat_merged.drop(columns=['LINKDT', 'LINKENDDT'])
```

In this section the final dataset with analyst coverage is created.

```python
#Ensure that permno is in the same format in both datasets for the merge
compustat_merged['permno'] = compustat_merged['permno'].astype(int)


#Sort the datasets by dates 
compustat_merged = compustat_merged.sort_values('filedate').reset_index(drop=True)
crsp_ibes_merged = crsp_ibes_merged.sort_values('MthCalDt').reset_index(drop=True)


#Perform the Merge 
final_dataset = pd.merge_asof(
    compustat_merged,          
    crsp_ibes_merged,          
    left_on='filedate',        
    right_on='MthCalDt',       
    by='permno',
    direction='backward',      
    tolerance=pd.Timedelta(days=40)       
)

#I create an artifical inner join (merge_asof has no inner join but I want to keep only observations where I have a match in both datasets)
#If no match found in CRSP_IBES data the MthCalDt will be NaN after the merge, so I can filter for non-NaN MthCalDt to keep only matched observations
final_dataset = final_dataset.dropna(subset=['MthCalDt']).reset_index(drop=True)
```

## **4. Constructing the Target Variables**

### 4.1. CAGR 

In this section the CAGR target variable for the predictive modeling task is constructed, being t+5 to t+10 Growth Rate calculations with fallback decisions, that is further specified in the thesis. First of all the function add_forward_cagr is built and applied to the final datasets. Then the ending horizons of the target variable are inspected to check if most observations are not affected by the fallback decisions. And this finally concludes with the previously developed case analysis of Apple to see if the target variable is constructed correctly.

```python
def add_forward_cagr(d, look, name, start=5, end_max=10, end_min=8, year_col='fyear'):
      """
      Forward geometric CAGR from t+start to t+end (end_max preferred down to end_min as a fallback),
      Adds y_{name}_{start}_{end_max} as a column to the dataset d.
      """

      #Full year column for the lookups
      fy = d[year_col].values

      # Retrieve the level variable (e.g. EBITDA, FCFF, Sales) at horizon h by matching each observation's (gvkey, fyear + h)
      def lag(h):
          return look.reindex(list(zip(d['gvkey'], fy + h))).values

      #Get the starting value of the variable at t+start and the ending value of the variable at t+end_max, t+end_max-1, ..., t+end_min
      x_start = lag(start)
      #Intialize the x_end (end variable) and end_use (end horizon) arrays with NaN values with the length of number of observations. Filled in the next loop with the first available value of x_end from t+end_max down to t+end_min (fallback)
      x_end   = np.full(len(d), np.nan)
      end_use = np.full(len(d), np.nan)

      #Iterate trough the end horizons starting from t+10 down to t+8.
      for h in range(end_max, end_min - 1, -1):
          v = lag(h)

          #If the end vector entry is empty (NaN) and the current lag vector entry is not empty (not NaN), I fill in the end vector entry with the current vector entry and record the horizon used in the end_use vector. 
          take = np.isnan(x_end) & ~np.isnan(v)
          x_end[take], end_use[take] = v[take], h

      #Calcularte the span between the start and end horizons & validity conditions
      span  = end_use - start
      valid = (x_start > 0) & (x_end > 0) & (span > 0)

      #Actually CAGR calculations of the formula in the thesis (with error handling for invalid or divide by zero cases)
      with np.errstate(invalid='ignore', divide='ignore'):
          cagr = (x_end / x_start) ** (1.0 / span) - 1

      #Add the columns to the dataset with the name of the variable and the horizon used for the end value
      col = f'{name}_cagr'
      d[col]          = np.where(valid, cagr, np.nan)
      d[f'{col}_end'] = np.where(valid, end_use, np.nan)
      return d
```

```python
#Ensure that the fyear column is numeric for the CAGR calculations
final_dataset['fyear'] = pd.to_numeric(final_dataset['fyear'], errors='coerce')

#For all the target variables and corresponding firm-year lookups, I calculate the forward CAGR and append the calculated values to the final dataset.
for nm, lk in y_lookups.items():
  final_dataset = add_forward_cagr(final_dataset, lk, nm)
  
# FCFF level 
final_dataset['fcff'] = final_dataset['oibdp'] - final_dataset['capx']

#As a sanity check I can look at the number of non-missing values for the calculated CAGR variables in the final dataset.
final_dataset[['ebitda_cagr', 'sale_cagr', 'fcff_cagr']].notna().sum()
```

There are clean observations ranging from 19.285 to 32.524 depending on the proxy variable in the dataset with analyst coverage.

### 4.2. Least Squares Growth 

Following Gao & Wu (2014/15) and Dechow, Hutton and Sloan (2010), a Least Squares approach is also incorporated as a second construction of the long run growth rate. The exact procedure is described in the thesis. Similar to the previous section, a function is built to calculate the least squares growth rate and applied to the final datasets.

```python
def add_lsq_growth(d, look, name, start=5, end=10, min_obs=4, last_min=8, year_col='fyear'):

    """ Forward geometric CAGR from t+start to t+end (end_max preferred down to end_min as a fallback),
    Adds y_{name}_{start}_{end_max} as a column to the dataset d.
    """

    #Full year column for the lookups
    fy = d[year_col].values

    #Initialize a array with the range of years from 5 to 10 and convert it to float type for calculations  
    taus = np.arange(start, end + 1).astype(float)

    # Retrieve the level variable (e.g. EBITDA, FCFF, Sales) at horizon h by matching each observation's (gvkey, fyear + h) like in the CAGR function. 
    #But this time I directly create a matrix M with the values of the variable at each horizon for each observation. Each row corresponds to an observation and each column corresponds to a horizon from 5 to 10.
    M = np.column_stack([look.reindex(list(zip(d['gvkey'], fy + int(h)))).values for h in taus])

    
    with np.errstate(invalid='ignore', divide='ignore'):
        #Calculate the log of the values in M only where they are positive. If a value is not positive, I set it to NaN to avoid taking the log of non-positive numbers.
        Y = np.where(M > 0, np.log(np.where(M > 0, M, np.nan)), np.nan)

        #Initialize a matrix T with the same shape as Y, where each row contains the taus (5, 6, 7, 8, 9, 10) values.
        T = np.tile(taus, (len(d), 1))

        #If a specific value in Y is NaN, I also set the corresponding value in T to NaN. This makes sure that I only consider valid pairs of (tau, log(value)) for the regression calculations.
        T[np.isnan(Y)] = np.nan

        #Number of non-NaN observations for each row (observation) in Y. This is used to check if I have enough data points to perform the regression.
        n = (~np.isnan(Y)).sum(axis=1)
        #Validity check
        ok = n >= 2  


        # Get the OLS coefficients for each observation 

        #Calculate the mean of T and Y for each observation, ignoring NaN values. If an observation does not have enough valid data points (less than 2), I set the mean to NaN.
        t_mean = np.where(ok, np.nanmean(np.where(ok[:, None], T, 0.0), axis=1), np.nan)
        y_mean = np.where(ok, np.nanmean(np.where(ok[:, None], Y, 0.0), axis=1), np.nan)

        #Calculate the covariance between T and Y for each observation, ignoring NaN values. The covariance is calculated as the sum of the product of the deviations of T and Y from their respective means.
        cov = np.nansum((T - t_mean[:, None]) * (Y - y_mean[:, None]), axis=1)
        var = np.nansum((T - t_mean[:, None]) ** 2, axis=1)

        #Get the last horizon non-NaN value  for each observation. 
        last = np.nanmax(np.where(np.isnan(Y), -np.inf, T), axis=1)

        #Validity check to see if each observation has enough non-NaN Values and if the last non-NaN value is at least last_min and if the variance is positive. If all these conditions are met, I consider the observation valid for regression.
        valid = (n >= min_obs) & (last >= last_min) & (var > 0)

        #Calculate the slope of the regression line for each observation. The slope is calculated as the covariance divided by the variance. If an observation is not valid, I set the slope to NaN.
        slope = np.where(valid, cov / np.where(var > 0, var, np.nan), np.nan)

    #Add the LSQ Growth and the number of observations for the regression to the Dataset
    col = f'{name}_lsq'
    #Since the values are logged I have to exponentiate the slope to get the geometric-equivalent annual growth rate and append it to the dataset
    d[col] = np.exp(slope) - 1                        
    d[f'{col}_nobs'] = np.where(valid, n, np.nan)
    return d
```

After defining the complex function, it is applied to the Dataset with Analyst Coverage.

```python
#For all the target variables and corresponding firm-year lookups, I calculate the forward LSQ and append the calculated values to the final dataset.
for nm, lk in y_lookups.items(): 
    final_dataset = add_lsq_growth(final_dataset, lk, nm)

#As a sanity check I can look at the number of non-missing values for the calculated LSQ  in the final dataset.
print(final_dataset[['ebitda_lsq', 'sale_lsq', 'fcff_lsq']].notna().sum())
```

### 6.6. Cash Flow Life Cycle

Following Dickinson (2011), the Cash Flow Life Cycle features are constructed in this section, which is explained in more detail in the thesis. The procedure is as follows: in a first step a dictionary with the different life cycle stages suggested by Dickinson is created. Then the dictionary items are iterated through and the life cycle stage is assigned to each firm-year pair in the final datasets.

```python
### Firm Cash Flow Life Cycle Dickinson:
dickinson_lifecycle = {'Introduction': [-1, -1,  1],
                       'Growth':       [ 1, -1,  1],
                       'Maturity':     [ 1, -1, -1],
                       'Shake-out':    [-1, -1, -1],
                       'Shake-out_2':  [ 1,  1,  1],
                       'Shake-out_3':  [ 1,  1, -1],
                       'Decline':      [-1,  1,  1],
                       'Decline_2':    [-1,  1, -1]}

#Initialize an empty lifecycle column in the final dataset
final_dataset['lifecycle'] = None

# Stage and 3 Conditions as a Tuple in a For Loop
for stage, (s_o, s_i, s_f) in dickinson_lifecycle.items():
    #mask is a True/False column true when all 3 conditions are met. np.sign function give 1.0 for positive or zero values, -1.0 for negative values.
    mask = (np.sign(final_dataset['oancf']) == s_o) & (np.sign(final_dataset['ivncf']) == s_i) & (np.sign(final_dataset['fincf']) == s_f)
    #For all observations where mask is True, assign the first part of the stage name (before the underscore) to the 'lifecycle' column in final_dataset
    final_dataset.loc[mask, 'lifecycle'] = stage.split('_')[0]

#Same procedure for the final_dataset_2 and final_dataset_3

final_dataset_2['lifecycle'] = None

for stage, (s_o, s_i, s_f) in dickinson_lifecycle.items():
    mask = (np.sign(final_dataset_2['oancf']) == s_o) & (np.sign(final_dataset_2['ivncf']) == s_i) & (np.sign(final_dataset_2['fincf']) == s_f)
    final_dataset_2.loc[mask, 'lifecycle'] = stage.split('_')[0]

final_dataset_3['lifecycle'] = None

for stage, (s_o, s_i, s_f) in dickinson_lifecycle.items():
    mask = (np.sign(final_dataset_3['oancf']) == s_o) & (np.sign(final_dataset_3['ivncf']) == s_i) & (np.sign(final_dataset_3['fincf']) == s_f)   
    final_dataset_3.loc[mask, 'lifecycle'] = stage.split('_')[0]
```

### 6.7. Industry Features

As suggested by previous empirical studies, industry features could potentially play a crucial role in the predictive modeling task. The industry classifier in this setting are the two first digits of the SIC Code. However, one cannot just look at the final dataset to construct the industry features at a given point in time. This is because of the filtering decisions made in the data cleaning process, potentially losing firms in a given industry. Therefore the industry features are created from the compustat funda dataset and, for the profitability features, fallback decisions are also incorporated to get the most complete industry features possible. To avoid time leakage concerns, lagged industry features (t-1) are used for the predictive modeling task.

These are the industry features constructed in this section (further details and formulas are provided in the Appendix of the thesis):

- Herfindahl-Hirschman Index (HHI)
- Industry-Median Profitability (ROA, ROE, OPM)
- Industry Durable Dummy

```python
#Construct the SIC2 industry classifier
final_dataset['sic2'] = final_dataset['sich'].astype(int) // 100
final_dataset_2['sic2'] = final_dataset_2['sich'].astype(int) // 100
final_dataset_3['sic2'] = final_dataset_3['sich'].astype(int) // 100
```

```python
#HHI per Industry Year

#Construct a new DataFrame funda_h with the relevant columns from compustat_funda,
funda_h = compustat_funda[['gvkey', 'fyear', 'sich', 'sic', 'sale']].dropna(subset=['fyear']).copy()
funda_h['sic2'] = funda_h['sich'].fillna(funda_h['sic']).astype(int) // 100
funda_h = funda_h[funda_h['sale'] > 0].drop_duplicates(['gvkey', 'fyear'], keep='last')

#Calculate the squared market share of each firm within its industry-year group. 
funda_h['share_sq'] = (funda_h['sale'] / funda_h.groupby(['sic2', 'fyear'])['sale'].transform('sum')) ** 2

#Sum the squared market shares to get the Herfindahl-Hirschman Index (HHI) for each industry-year group.
hhi = (funda_h.groupby(['sic2', 'fyear'])[['share_sq']]
       .sum()
       .reset_index()
       .rename(columns={'share_sq': 'hhi'}))

#Time Leakage Protection: Shift the fyear in the HHI DataFrame by 1 
hhi['fyear'] = hhi['fyear'] + 1

#Merge the new feature HHI into the final datasets 
final_dataset = final_dataset.merge(hhi, on=['sic2', 'fyear'], how='left')
final_dataset_2 = final_dataset_2.merge(hhi, on=['sic2', 'fyear'], how='left')
final_dataset_3 = final_dataset_3.merge(hhi, on=['sic2', 'fyear'], how='left')
```

```python
# Lagged Industry-Median Profitability (ROA, ROE, OPM) per Year 


# (1) winsorize the INPUTS per fyear (1/99) so one micro-sales firm can't own a thin cell's median
# (2) hierarchical fallback: 2-digit SIC → 1-digit SIC → all-market whenever a cell has < MIN_N firms
# (3) t-1 → t leakage lag unchanged

MIN_N = 5   # min firms per industry-year cell before falling back to a coarser grouping


#Create a new DataFrame funda_p with the relevant columns from compustat_funda, dropping rows with missing fyear values and creating the industry classifiers
funda_p = compustat_funda[['fyear', 'sich', 'sic', 'ib', 'at', 'ceq', 'oibdp', 'sale']].dropna(subset=['fyear']).copy()
funda_p['sic_full'] = funda_p['sich'].fillna(funda_p['sic']).astype(int)
funda_p['sic2'] = funda_p['sic_full'] // 100    # 2-digit industry (as before)
funda_p['sic1'] = funda_p['sic_full'] // 1000   # 1-digit division (coarse fallback)

#Create the profitability ratios ROA, ROE, and operating margin (op_margin) for each firm-year observation in funda_p. 
funda_p['ROE'] = (funda_p['ib'] / funda_p['ceq'] * 100).where(funda_p['ceq'] > 0).fillna(0)
funda_p = funda_p[funda_p['sale'] > 0]
funda_p['ROA'] = funda_p['ib'] / funda_p['at'] * 100
funda_p['op_margin'] = funda_p['oibdp'] / funda_p['sale'] * 100

# Winsorization of the inputs, per fyear
for c in ['ROA', 'ROE', 'op_margin']:
    grp = funda_p.groupby('fyear')[c]
    funda_p[c] = funda_p[c].clip(grp.transform(lambda x: x.quantile(0.01)),
                                 grp.transform(lambda x: x.quantile(0.99)))

#Create function that calculates the median of ROA, ROE, and op_margin and size for each group defined by the keys (e.g., ['sic2', 'fyear'] and adds a suffix to the column names.
def level_median(keys, suffix):
    g = funda_p.groupby(keys)[['ROA', 'ROE', 'op_margin']]
    out = g.median()
    out['n'] = g.size()
    return out.add_suffix(suffix).reset_index()

#Construct the Main & Fallback Values 
m2 = level_median(['sic2', 'fyear'], '2')   
m1 = level_median(['sic1', 'fyear'], '1')
m0 = level_median(['fyear'], '0')           

#For all industry year cells, I merge the main and fallback values together to create a new DataFrame called cells. 
cells = (funda_p[['sic2', 'sic1', 'fyear']].drop_duplicates()
         .merge(m2, on=['sic2', 'fyear'], how='left')
         .merge(m1, on=['sic1', 'fyear'], how='left')
         .merge(m0, on=['fyear'],         how='left'))

#If the enough firms are in the 2-digit SIC cell, I use the 2-digit SIC median. If not, I check if there are enough firms in the 1-digit SIC cell and use that median. If neither has enough firms, we fall back to the all-market median. 
use2 = cells['n2'] >= MIN_N
use1 = (~use2) & (cells['n1'] >= MIN_N)


industry_medians = cells[['sic2', 'fyear']].copy()

#Iterate trough the profitability ratios (ROA, ROE, op_margin) and create new columns in the industry_medians DataFrame with the appropriate median values based on the hierarchical fallback logic.
for base, out in [('ROA', 'roa_industry_median'),
                  ('ROE', 'roe_industry_median'),
                  ('op_margin', 'opm_industry_median')]:
    industry_medians[out] = np.where(use2, cells[base + '2'],
                             np.where(use1, cells[base + '1'], cells[base + '0']))

#Time Leakage Protection: Shift the fyear in the industry_medians DataFrame by 1 
industry_medians['fyear'] = industry_medians['fyear'] + 1

#Merge the industry median profitability ratios into the final datasets
final_dataset   = final_dataset.merge(industry_medians,   on=['sic2', 'fyear'], how='left')
final_dataset_2 = final_dataset_2.merge(industry_medians, on=['sic2', 'fyear'], how='left')
final_dataset_3 = final_dataset_3.merge(industry_medians, on=['sic2', 'fyear'], how='left')
```

```python
#Indstury Durables vs No Durables Dummies from
#NAICS to SIC Conversion: 
#https://www.naics.com/naics-code-description/?code=423 https://www.naics.com/naics-code-description/?code=423
#https://www.naics.com/wp-content/uploads/2014/10/NAICS-to-SIC-Crosswalk.pdf

durable_sic2 = [24, 25, 32, 33, 34, 35, 36, 37, 38, 39]

final_dataset['type_durable']   = final_dataset['sic2'].isin(durable_sic2).astype(int)
final_dataset_2['type_durable'] = final_dataset_2['sic2'].isin(durable_sic2).astype(int)
final_dataset_3['type_durable'] = final_dataset_3['sic2'].isin(durable_sic2).astype(int)
```
